AIM - Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.

Step 1: Import Pandas Library

```python
# Step 1: Import the pandas library
import pandas as pd
```

Step 2: Create the Dataset

```python
# Step 2: Create sample data using a dictionary
data = {
    "ID": [1, 2, 3, 4, 5, 6, 7],
    "Name": ["Priya", "Amit", None, "Riya", "Ajay", "Vijay", "Jeni"],
    "Marks": [21.0, None, 20.0, 20.0, 22.0, 23.0, 28.0],
    "City": ["Surat", "Navsari", "Surat", "Bhavnagar", "Bharuch", None, None],
    "Age": [25, None, 5, 8, 7, None, None]
}
```

Step 3: Create DataFrame

```python
# Step 3: Convert the dictionary into a DataFrame
df = pd.DataFrame(data)
```

Step 4: Display the Dataset

```python
# Step 4: Display the complete dataset
print(df)
```

Step 5: Display First Five Rows

```python
# Step 5: Display the first five rows
df.head()
```

Step 6: Display Last Five Rows

```python
# Step 6: Display the last five rows
df.tail()
```

Step 7: Display Dataset Information

```python
# Step 7: Display information about the dataset
df.info()
```

Step 8: Display Statistical Information

```python
# Step 8: Display the statistical summary of numerical columns
df.describe()
```

Step 9: Display Shape of Dataset

```python
# Step 9: Display the number of rows and columns
df.shape
```

Step 10: Display Column Names

```python
# Step 10: Display the names of all columns
df.columns
```

Step 11: Display Data Types

```python
# Step 11: Display the data type of each column
df.dtypes
```

Step 12: Check Missing Values

```python
# Step 12: Count missing values in each column
df.isnull().sum()
```

Step 13: Fill Missing Values in Marks

```python
# Step 13: Replace missing marks with the average marks
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

Step 14: Fill Missing Values in Age

```python
# Step 14: Replace missing age with the average age
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

Step 15: Fill Missing Names

```python
# Step 15: Replace missing names with "Unknown"
df["Name"] = df["Name"].fillna("Unknown")
```

Step 16: Fill Missing City

```python
# Step 16: Replace missing city with "Unknown"
df["City"] = df["City"].fillna("Unknown")
```

Step 17: Check Duplicate Rows

```python
# Step 17: Check for duplicate records
df.duplicated()
```

Step 18: Count Duplicate Rows

```python
# Step 18: Count the total number of duplicate rows
df.duplicated().sum()
```

Step 19: Remove Duplicate Rows

```python
# Step 19: Remove duplicate records
df = df.drop_duplicates()
```

Step 20: Rename Column

```python
# Step 20: Rename the Marks column to Score
df.rename(columns={"Marks": "Score"}, inplace=True)
```

Step 21: Change Data Type

```python
# Step 21: Convert the Age column to integer
df["Age"] = df["Age"].astype(int)
```

Step 22: Create New Column

```python
# Step 22: Create a new FinalScore column
# Add 5 bonus marks to the Score
df["FinalScore"] = df["Score"] + 5
```

Step 23: Convert Names to Uppercase

```python
# Step 23: Convert all names to uppercase
df["Name"] = df["Name"].str.upper()
```

Step 24: Convert City to Lowercase

```python
# Step 24: Convert all city names to lowercase
df["City"] = df["City"].str.lower()
```

Step 25: Remove Extra Spaces

```python
# Step 25: Remove leading and trailing spaces from city names
df["City"] = df["City"].str.strip()
```

Step 26: Sort the Dataset

```python
# Step 26: Sort the dataset by FinalScore in descending order
df = df.sort_values(by="FinalScore", ascending=False)
```

Step 27: Save the Cleaned Dataset

```python
# Step 27: Save the cleaned dataset as a CSV file
df.to_csv("cleaned_students.csv", index=False)

# Display success message
print("Cleaned dataset saved successfully.")
```

Step 28: Display Final Dataset

```python
# Step 28: Display the final cleaned dataset
print(df)
```
