Step1: Display the Dataset

```python
# Import the Pandas library for data manipulation and analysis
import pandas as pd
```

Step 2: Create the Dataset

```python
# Create a dictionary containing structured employee details
data = {
     "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Neha", "Vivek", "Pooja"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance", "IT", "HR"],
    "City": ["Surat", "Navsari", "Surat", "Valsad", "Surat", "Navsari", "Valsad", "Surat"],
     "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience": [2, 1, 5, 4, 2, 6, 8, 3]
}
```

Step 3: Create DataFrame

```python
# Convert the dictionary into a tabular Pandas DataFrame structure
df = pd.DataFrame(data)

# Print the final DataFrame to the console to view the table
print(df)
```

Step 4: Filter Data

Display Employees From IT Department

```python
# Filter and display only the employees who work in the IT department
df[df["Department"] == "IT"]
```

Display Employees with Salary Greater Than 50,000

```python
#Display the Employees whose salary is greater than 50,000
df[df["Salary"]>50000]
```

Display Employee From Surat

```python
#Display the Employees who are from "Surat"
df[df["City"]=="Surat"]
```

Apply Multiple Conditions

```python
#Display the Employees whose Department is "IT" and Salary is greater than 50,000
df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

Step 5: Sort Data

Sort Salary in Ascending Order

```python
# Sort rows based on the 'Salary' column (defaults to ascending order)
df.sort_values(by="Salary")
```

Sort Salary in Descending Order

```python
# Sort rows by 'Salary' and set ascending to False for a descending order result
df.sort_values(by="Salary", ascending=False)
```

Sort by Department and Salary

```python
# Sort rows using a list of columns and map corresponding boolean flags for sort order
df.sort_values(by=["Department", "Salary"], ascending=[True, False])
```

Step 6: Perform Aggregation


Calculate Total Salary

```python
# Calculate the total sum of the Salary column
df["Salary"].sum()
```

Calculate Average Salary

```python
#Calculate the average salary of all employees
df["Salary"].mean()
```

Find Highest Salary

```python
#Find the Highest salary from the Salary column
df["Salary"].max()
```

Find Lowest Salary

```python
# Find the lowest value in the Salary column
df["Salary"].min()
```

Count Employees

```python
#Count the number of non-empty values in the Emp_ID column
df["Emp_ID"].count()
```

Step7: Perform GroupBy Operations

Average Salary Department-wise

```python
#Calculate the Average Salary for each Department
df.groupby("Department")["Salary"].mean()
```

Minimum Salary Department-wise

```python
#Calculate the Minimum Salary for each Department
df.groupby("Department")["Salary"].min()
```

Count Total Employees Department-wise

```python
# Group data by the 'Department' column and compute the count of non-null 'Emp_ID' values per group
df.groupby("Department")["Emp_ID"].count()
```

Group by City

```python
#Group the data by "City" and Calculate the Average Salary for each City
df.groupby("City")["Salary"].mean()
```

Total Salary Deparetment-wise

```python
# Total salary sum for each department
df.groupby("Department")["Salary"].sum()
```

Step 8: Multiple Aggregation Functions

```python
#Group the data by "Department" and Calculate multiple statistics for the Salary column
df.groupby("Department")["Salary"].agg(["count","sum","mean","min","max"])
```

Step 9: Save the Result

```python
#Save the DataFrame as a CSV file named "Employee_Data.csv" and index=False prevents pandas from saving the row index as an extra column after that print success message
df.to_csv("Employee_Data.csv",index=False)
print("Dataset Saved Successfully.")
```
