Step 1: Import Pandas Library

```python
# Step 1: Import the pandas library
import pandas as pd
```

Step 2: Create Sample Data

```python
# Step 2: Create sample student data
data={
    "ID":[101,102,103,104,105,106,107],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Krisha","Harry"],
    "Age":[20,21,19,20,22,24,28],
    "City":["Surat","Navsari","Valsad","Bharuch","Surat","Gandhinagar","Ahmedabad"],
    "Marks":[85,92,78,88,81,84,90]
}
```

Step 3: Create a DataFrame

```python
# Step 3: Create a DataFrame from the sample data
df=pd.DataFrame(data)

# Display the DataFrame
df
```

Step 4: Save Data as CSV File

```python
# Step 4: Save the DataFrame as a CSV file
df.to_csv("students.csv",index=False)

# Display success message
print("csv file created successfully.")
```

Step 5: Save Data as Excel File

```python
# Step 5: Save the DataFrame as an Excel file
df.to_excel("students.xlsx", index=False)

# Display success message
print("Excel file created successfully.")
```

Step 6: Save Data as JSON File

```python
# Step 6: Save the DataFrame as a JSON file
df.to_json("students.json", orient="records", indent=4)


# Display success message
print("JSON File created successfully.")
```

Step 7: Read the CSV File
