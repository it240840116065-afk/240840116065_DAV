Aim: Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).

Step 1: Create a Line Chart
import Libraries

```python
#import the pandas library.
import pandas as pd

#imoprt the mutplotlib library for creating graph
import matplotlib.pyplot as plt
```

```python
data ={
    "Month":['January', 'February', 'March', 'April', 'May', 'June',
              'July', 'August', 'September', 'October', 'November', 'December'],
"Product":['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet'],
    "Region":['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
"Sales": [120000, 135000, 150000, 145000, 160000, 175000,
                180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40]
    }

df = pd.DataFrame(data)

print(df)
```

```python
# Line chart
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

Step 2: January to December Sales

```python
# Show January and December sales
print("January Sales:", df["Sales"][0])
print("December Sales:", df["Sales"][11])
```

Step 3: Pie Chart

```python
# Product wise sales
x = df.groupby("Product")["Sales"].sum()
plt.pie(x, labels=x.index, autopct="%1.1f%%")
plt.title("Sales by Product")
plt.show()
# Show product sales
print(x)

```

Step 4: Create Histogram


```python
# Histogram of sales
plt.hist(df["Sales"], bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()

# Show sales distribution
print(df["Sales"].describe())

```

Step 5: Units Sold and Total Sales

```python
# Scatter plot
plt.scatter(df["Units_Sold"], df["Sales"])
plt.xlabel("Units Sold")
plt.ylabel("Sales")
plt.title("Units Sold vs Sales")
plt.show()
```

Step 6: Pie Chart – Region Sales

```python
# Region wise sales
x = df.groupby("Region")["Sales"].sum()
plt.pie(x, labels=x.index, autopct="%1.1f%%")
plt.title("Sales by Region")

plt.show()
```

Step 7: Save Dataset

```python
# Save data as CSV
df.to_csv("sales.csv", index=False)
print("File saved")
```
