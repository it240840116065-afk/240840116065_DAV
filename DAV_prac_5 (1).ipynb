Aim: Analyze datasets using correlation, covariance, outlier detection, and distribution analysis.

Step 1:  Import required libraries

```python
#: Import required libraries for data analysis and visualization import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Step 2: Create a Dataset

```python
# Define the student data dictionary with relevant academic metrics
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114, 115],

    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9, 3, 4, 5, 6, 7, 10, 1],

    "Attendance": [65, 70, 75, 80, 82, 85, 90, 92, 68, 74, 78, 84, 88, 95, 40],

    "Assignment_Score": [55, 60, 65, 70, 72, 78, 82, 88, 62, 67, 71, 76, 80, 95, 35],

    "Final_Score": [50, 55, 62, 68, 72, 78, 82, 90, 58, 63, 69, 75, 81, 98, 25]
}

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the complete DataFrame
print(df)
```

Step 3: Inspect Dataset

```python
# Display the first five records of the dataset
print(df.head())
```

```python
#Display information about the dataset

print("Dataset Information:")
df.info()
```

```python
#Display number of rows and columns

rows, columns = df.shape

print("Number of Rows:", rows)
print("Number of Columns:", columns)
```

```python
#Generate basic descriptive statistics

print("Basic Descriptive Statistics:")
print(df.describe())
```

Step 4: Calculate Correlation

```python
#Calculate correlation between Study Hours and Final Score

correlation = df["Study_Hours"].corr(df["Final_Score"])

print("Correlation between Study Hours and Final Score:")
print(correlation)
```

Step 5: Calculate Correlation Matrix

```python
#Calculate correlation among all numerical variables

correlation_matrix = df.corr(numeric_only=True)

# Display the correlation matrix
print("Correlation Matrix:")
print(correlation_matrix)
```

Step 6: Visualize Correlation Using Scatter Plot

```python
# Create a scatter plot
# to visualize the relationship between Study Hours and Final Score

plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Final_Score"])

plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")

plt.show()
```

Step 7: Calculate Covariance

```python
# Calculate covariance between Study Hours and Final Score

covariance = df["Study_Hours"].cov(df["Final_Score"])

# Display covariance
print("Covariance between Study Hours and Final Score:")
print(covariance)
```

Step 8: Calculate Covariance Matrix

```python
# Calculate Covariance Matrix

# Select numerical columns
numerical_columns = df.select_dtypes(include="number")

# Calculate covariance matrix
covariance_matrix = numerical_columns.cov()

# Display covariance matrix
print("Covariance Matrix:")
print(covariance_matrix)
```

Step 9: Detect Outliers Using IQR

```python
# Detect Outliers Using IQR

# Select Final_Score column
final_score = df["Final_Score"]

# Calculate Q1 and Q3
Q1 = final_score.quantile(0.25)
Q3 = final_score.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Display the values
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

Step 10: Identify Outliers

```python
# Select records below the lower bound
# or above the upper bound
outliers = df[
    (df["Final_Score"] < lower_bound) |
    (df["Final_Score"] > upper_bound)
]

# Display the outlier records
print("Detected Outlier Records:")
print(outliers)
```

Step 11: Count Outliers

```python
# Count the total number of outliers

outlier_count = len(outliers)

print("Total Number of Outliers:")
print(outlier_count)
```

Step 12: Visualize Outliers Using Box Plot

```python
# Visualize Outliers Using Box Plot

plt.figure(figsize=(8, 5))

plt.boxplot(df["Final_Score"])

plt.xlabel("Final Score")
plt.ylabel("Score")
plt.title("Box Plot of Final Score")

plt.show()
```

Step 13: Create a Histogram

```python
# Create a histogram to observe the distribution of Final Score

# Create the histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Final_Score"], bins=5)

# Add labels and title
plt.xlabel("Final Score")
plt.ylabel("Frequency")
plt.title("Distribution of Final Score")

# Display the histogram
plt.show()
```

Step 14: Analyze Distribution Using Skewness

```python
# Calculate the skewness of Final Score

# Calculate skewness
skewness = df["Final_Score"].skew()

# Display the skewness value
print("Skewness of Final Score:")
print(skewness)
```

Step 15: Calculate Kurtosis

```python
# Calculate the kurtosis of Final Score

# Calculate kurtosis
kurtosis = df["Final_Score"].kurt()

# Display the kurtosis value
print("Kurtosis of Final Score:")
print(kurtosis)
```

Step 16: Distribution of Study Hours

```python
# Create a histogram for Study Hours

# Create the histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Study_Hours"], bins=5)

# Add labels and title
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.title("Distribution of Study Hours")

# Display the histogram
plt.show()
```

Step 17: Compare Variables Using Box Plots

```python
# Compare variables using box plots

# Select numerical variables for box plot
numerical_data = df[
    ["Study_Hours", "Attendance", "Assignment_Score", "Final_Score"]
]

# Create box plots
plt.figure(figsize=(10, 6))
numerical_data.boxplot()

# Add title and labels
plt.title("Comparison of Numerical Variables Using Box Plots")
plt.xlabel("Variables")
plt.ylabel("Values")

# Display the box plots
plt.show()
```
